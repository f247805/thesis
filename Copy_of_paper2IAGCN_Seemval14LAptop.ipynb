```python
import pandas as pd
import spacy
import torch
# 1. Load BERT tokenizer and model
from transformers import BertTokenizerFast, AutoTokenizer, AutoModel # Make sure to import BertTokenizerFast

# Initialize tools
# nlp = spacy.load("en_core_web_sm")  # You might need to reload this after upgrading torchvision
nlp = spacy.load("en_core_web_sm") # Load en_core_web_sm

tokenizer = BertTokenizerFast.from_pretrained("bert-base-uncased")
bert_model = AutoModel.from_pretrained("bert-base-uncased")
# ... (Rest of your code)
# Helper function to calculate dependency distance
def get_dep_distance(token1, token2):
    # Traverse up to root
    def get_ancestors(token):
        ancestors = []
        while token.head != token:
            ancestors.append(token)
            token = token.head
        ancestors.append(token)  # Add the root
        return ancestors

    ancestors1 = get_ancestors(token1)
    ancestors2 = get_ancestors(token2)

    # Find lowest common ancestor (LCA)
    lca = None
    for ancestor in ancestors1:
        if ancestor in ancestors2:
            lca = ancestor
            break

    if lca is None:
        return len(ancestors1) + len(ancestors2)  # No common ancestor
    else:
        return ancestors1.index(lca) + ancestors2.index(lca)



def create_adj_matrix(text):
     # Get BERT tokenization
    # encoding = bert_tokenizer(text, return_offsets_mapping=True) # 'bert_tokenizer' is not defined, change to 'tokenizer'
    encoding = tokenizer(text, return_offsets_mapping=True) # Use the defined 'tokenizer' variable
    bert_tokens = tokenizer.convert_ids_to_tokens(encoding.input_ids)

    # Get spaCy parse
    doc = nlp(text)

    # Create alignment map between spaCy and BERT tokens
    alignment_map = []
    for token in doc:
        start, end = token.idx, token.idx + len(token.text)
        aligned = [
            i for i, (tok_start, tok_end) in enumerate(encoding.offset_mapping)
            if tok_start >= start and tok_end <= end
        ]
        alignment_map.append(aligned)

    # Build adjacency matrix
    size = len(bert_tokens)
    matrix = torch.eye(size)

    # Add dependency relations
    for token in doc:
        head = token.head
        if token.i != head.i:  # Skip self-connections
            for src in alignment_map[token.i]:
                for tgt in alignment_map[head.i]:
                    matrix[src, tgt] = 1
                    matrix[tgt, src] = 1  # Undirected

    return matrix.numpy()
# Aspect token span finding
def char_to_token_positions(row):
    encoding = tokenizer(row["Sentence"], return_offsets_mapping=True)
    aspect_start_char = row["from"]
    aspect_end_char = row["to"]

    # Tokenize aspect
    aspect_tokens = tokenizer.tokenize(row["Aspect Term"])

    # Find token positions using character offsets
    aspect_start_token = None
    aspect_end_token = None
    for idx, (start, end) in enumerate(encoding.offset_mapping):
        if start <= aspect_start_char < end:
            aspect_start_token = idx
        if start < aspect_end_char <= end:
            aspect_end_token = idx
            break

    # Fallback 1: Token sequence matching
    if aspect_start_token is None or aspect_end_token is None:
        tokens = tokenizer.tokenize(row["Sentence"])
        for i in range(len(tokens) - len(aspect_tokens) + 1):
            if tokens[i:i+len(aspect_tokens)] == aspect_tokens:
                aspect_start_token, aspect_end_token = i, i+len(aspect_tokens)
                break

    # Fallback 2: Default to first token
    if aspect_start_token is None or aspect_end_token is None:
        aspect_start_token = 0
        aspect_end_token = min(1, len(tokens))

    # Ensure correct span
    if (aspect_end_token - aspect_start_token) < len(aspect_tokens):
        aspect_end_token = aspect_start_token + len(aspect_tokens)

    return aspect_start_token, aspect_end_token

def calculate_srd(row):
    try:
        doc = nlp(row["Sentence"])
        distances = []

        start = row["aspect_start_token"]
        end = row["aspect_end_token"]

        if pd.isna(start) or pd.isna(end):
            return [0] * len(doc)
        start = int(start)
        end = int(end)

        if start >= end or end > len(doc):
            return [0] * len(doc)

        aspect_tokens = list(doc[start:end])

        for token in doc:
            # Check if token is part of aspect
            if start <= token.i < end:
                distances.append(0)
                continue

            # Calculate minimum dependency distance to any aspect token
            min_dist = min(get_dep_distance(token, aspect_token) for aspect_token in aspect_tokens)
            distances.append(min_dist)

        return distances

    except Exception as e:
        print(f"Error in row {row.name}: {e}")
        return [0] * len(row["Sentence"].split())


# ✅ Master function
def preprocess_data(df):
    # Add aspect token positions
    df[["aspect_start_token", "aspect_end_token"]] = df.apply(
        char_to_token_positions,
        axis=1,
        result_type="expand"
    )

    # Add adjacency matrix
    df["adj_matrix"] = df["Sentence"].apply(create_adj_matrix)

    # Add SRD distances
    df["srd_weights"] = df.apply(calculate_srd, axis=1)

    return df

# Now run on your train data
# Example:
# train_data = pd.read_csv('your_train_file.csv')
# Filter conflict samples BEFORE preprocessing
train_data = train_data[train_data['polarity'] != 'conflict']


df = preprocess_data(train_data)
```

```python
from torch.utils.data import Dataset
import torch

class AspectDataset(Dataset):
    def __init__(self, df, max_len=128):
        self.df = df
        self.max_len = max_len
        self.label_map = {"negative": 0, "neutral": 1, "positive": 2}
    def __len__(self):
            return len(self.df)
    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        # Use alpha_weights instead of srd_weights
        # Access alpha_weights from the row:
        alpha_weights = torch.tensor(
            row["alpha_weights"][:self.max_len] + [0] * (self.max_len - len(row["alpha_weights"]))
            #row.get("alpha_weights", [0] * self.max_len)[:self.max_len] # or use get and default
        )
        # Convert to integers
        try:
            start = int(row["aspect_start_token"])
            end = int(row["aspect_end_token"])
        except:
            start, end = 0, 1

        # Clamp values
        start = max(0, min(start, self.max_len-1))
        end = max(start+1, min(end, self.max_len))

        aspect_mask = torch.zeros(self.max_len)
        aspect_mask[start:end] = 1

        # Tokenization with BERT tokenizer
        encoding = tokenizer(
            row["Sentence"],
            padding="max_length",
            truncation=True,
            max_length=self.max_len,
            return_tensors="pt"
        )

        # Aspect mask
        aspect_mask = torch.zeros(self.max_len)
        start = min(int(row["aspect_start_token"]), self.max_len-1)
        end = min(int(row["aspect_end_token"]), self.max_len)
        aspect_mask[start:end] = 1


        # Adjacency matrix
        adj_matrix = torch.tensor(row["adj_matrix"])
        if adj_matrix.shape[0] < self.max_len:
            pad = self.max_len - adj_matrix.shape[0]
            adj_matrix = torch.nn.functional.pad(adj_matrix, (0, pad, 0, pad))

        # SRD weights
        srd = torch.tensor(row["SRD"][:self.max_len] + [0]*(self.max_len - len(row["SRD"])))

        return {
            "input_ids": encoding["input_ids"].squeeze(),
            "attention_mask": encoding["attention_mask"].squeeze(),
            "adj_matrix": adj_matrix[:self.max_len, :self.max_len],
            "aspect_mask": aspect_mask,
            "srd_weights": srd,
            "alpha_weights": alpha_weights,  # Replace srd_weights
            "labels": torch.tensor(self.label_map[row["polarity"]], dtype=torch.long)
        }

# Create datasets
# 2. Calculate SRD
train_data["SRD"] = train_data.apply(calculate_srd, axis=1)
test_data["SRD"] = test_data.apply(calculate_srd, axis=1)
# ✅ Add this line after calculating SRD weights

# Ensure alpha_weights is calculated for both train_data and test_data:
train_data["alpha_weights"] = train_data["srd_weights"].apply(
    lambda srd_list: [1 / (1 + d) for d in srd_list]
)
test_data["alpha_weights"] = test_data["srd_weights"].apply(
    lambda srd_list: [1 / (1 + d) for d in srd_list]
)
train_dataset = AspectDataset(train_data)
test_dataset = AspectDataset(test_data)
```

```python
class GCNLayer(nn.Module):
    def __init__(self, in_features, out_features):
        super(GCNLayer, self).__init__()
        self.fc = nn.Linear(in_features, out_features)

    def forward(self, adj_matrix, features):
        # Apply adjacency matrix to features
        out = torch.matmul(adj_matrix, features)
        out = self.fc(out)
        return out
```

```python
class InteractiveAttention(nn.Module):
    def __init__(self, hidden_dim):
        super().__init__()
        # Context-to-Aspect attention parameters
        self.W_b = nn.Linear(hidden_dim, hidden_dim)
        self.U_b = nn.Linear(hidden_dim, hidden_dim)
        self.v_b = nn.Linear(hidden_dim, 1)

        # Aspect-to-Context attention parameters
        self.W_a = nn.Linear(hidden_dim, hidden_dim)
        self.U_a = nn.Linear(hidden_dim, hidden_dim)
        self.v_a = nn.Linear(hidden_dim, 1)

    def forward(self, context, aspect):
        # Context-to-Aspect attention (Equations 14-15)
        h_c = torch.tanh(self.W_b(context) + self.U_b(aspect).mean(1, keepdim=True))
        alpha = F.softmax(self.v_b(h_c), dim=1)
        D_r = (alpha * context).sum(1)

        # Aspect-to-Context attention (Equations 16-17)
        h_a = torch.tanh(self.W_a(aspect) + self.U_a(context).mean(1, keepdim=True))
        beta = F.softmax(self.v_a(h_a), dim=1)
        b_r = (beta * aspect).sum(1)

        # Return concatenated features (Equation 19)
        combined = torch.cat([D_r, b_r], dim=1)
        return combined  # Return only the tensor, not tuple
```

```python
class IAGCN(nn.Module):
    def __init__(self, vocab_size, embed_dim=300, hidden_dim=300, num_classes=3):
        super().__init__()
        # 1. Embedding Layer
      # 1. BERT Model (embedding_dim=768)
        self.bert = AutoModel.from_pretrained("bert-base-uncased")

        # 2. BiLSTM
        self.bilstm = nn.LSTM(
            768,  # BERT hidden size
            hidden_dim // 2,
            bidirectional=True,
            batch_first=True
        )


        # 3. GCN Layers with proper initialization
        self.gcn1 = nn.Linear(hidden_dim, hidden_dim)
        self.gcn2 = nn.Linear(hidden_dim, hidden_dim)

        # 4. Attention Mechanism
        self.attention = InteractiveAttention(hidden_dim)
        self.classifier = nn.Linear(2 * hidden_dim, num_classes)

        # 5. Classifier
        self.classifier = nn.Linear(2 * hidden_dim, num_classes)

        # 6. Activation and Dropout
        self.relu = nn.ReLU()
        self.dropout = nn.Dropout(0.1)
        self.gcn_drop = nn.Dropout(0.2)  # Additional GCN-specific dropout

    def forward(self, input_ids, adj_matrix, srd_weights, aspect_mask, alpha_weights):

        # 1. Embedding Layer
        bert_output = self.bert(input_ids).last_hidden_state

        # 2. BiLSTM
        lstm_out, _ = self.bilstm(bert_output)
        lstm_out = lstm_out * srd_weights.unsqueeze(-1)

        # 3. SRD-Weighted Adjacency Matrix
        adj_matrix = adj_matrix * srd_weights.unsqueeze(-1)  # (B, L, L)

        # 4. Add self-connections and normalize
        adj_matrix = adj_matrix * alpha_weights.unsqueeze(-1)   # Add self-loops
        degree_matrix = torch.sum(adj_matrix, dim=-1, keepdim=True)
        adj_matrix_norm = adj_matrix / (degree_matrix + 1e-6)

        # 5. GCN Processing with activations
        gcn_out = self.relu(torch.matmul(adj_matrix_norm, self.gcn1(lstm_out)))
        gcn_out = self.dropout(gcn_out)
        gcn_out = self.relu(torch.matmul(adj_matrix_norm, self.gcn2(gcn_out)))

        # 6. Aspect Feature Extraction
        aspect_feats = gcn_out * aspect_mask.unsqueeze(-1)
        c_avg = gcn_out.mean(1)
        a_avg = aspect_feats.mean(1)

        #D_r = (gcn_out * torch.sigmoid(gcn_out @ a_avg.unsqueeze(-1))).sum(1)
        #b_r = (aspect_feats * torch.sigmoid(aspect_feats @ c_avg.unsqueeze(-1))).sum(1)

        #return self.classifier(torch.cat([D_r, b_r], -1))
        # Attention
        aspect_feats = gcn_out * aspect_mask.unsqueeze(-1)
        attended = self.attention(gcn_out, aspect_feats)  # Unpack the tuple here

        return self.classifier(attended)  # Pass only attended_features to the classifier# 6. Aspect Feature Extraction
        aspect_feats = gcn_out * aspect_mask.unsqueeze(-1)

        # 7. Interactive Attention (now returns single tensor)
        attended = self.attention(gcn_out, aspect_feats)

        # 8. Final classification
        return self.classifier(attended)

    def predict(self, text, aspect_term):
        # Automatic preprocessing
            inputs = self._preprocess(text, aspect_term)

            # Prediction
            with torch.no_grad():
                outputs = self.model(**inputs)
                probs = torch.softmax(outputs, dim=1)

            return self._format_output(probs, text, aspect_term)
    def _preprocess(self, text, aspect_term):
            # 1. Automatic character position detection
            start = text.find(aspect_term)
            end = start + len(aspect_term)

            # 2. Handle tokenization and aspect masking
            encoding = self.tokenizer(text, return_tensors='pt', return_offsets_mapping=True)
            aspect_mask = self._create_aspect_mask(encoding, start, end)

            # 3. Dependency parsing
            adj_matrix = self._create_dep_matrix(text)

            # 4. SRD calculation
            srd_weights = self._calculate_srd(text, start, end)

            return {
                'input_ids': encoding.input_ids.to(self.device),
                'adj_matrix': adj_matrix.to(self.device),
                'srd_weights': srd_weights.to(self.device),
                'aspect_mask': aspect_mask.to(self.device)
            }
```

```python
import torch.optim as optim
from sklearn.metrics import f1_score
best_val_f1 = 0
patience = 2 # Define patience for early stopping
counter = 0  #
# Optimizer & Loss
optimizer = optim.AdamW(model.parameters(), lr=2e-5, weight_decay=1e-5)
criterion = nn.CrossEntropyLoss()

# Metrics
def calculate_metrics(preds, labels):
    acc = (preds == labels).float().mean()
    f1 = f1_score(labels.cpu(), preds.cpu(), average='weighted')
    return acc.item(), f1

best_val_f1 = 0
for epoch in range(20):
    if epoch == 3:
       model.bert.requires_grad_(True)
    # Training
    model.train()
    total_loss = 0
    all_preds = []
    all_labels = []

    for batch in train_loader:
        optimizer.zero_grad()

        inputs = batch['input_ids'].to(device)
        adj = batch['adj_matrix'].float().to(device)
        srd = batch['srd_weights'].float().to(device)
        mask = batch['aspect_mask'].float().to(device)
        labels = batch['labels'].to(device)
        alpha_weights = batch['alpha_weights'].float().to(device)  # Get alpha_weights from the batch
        outputs = model(inputs, adj, srd, mask, alpha_weights) # Pass alpha_weights to the model
    # ... (rest of the training loop)
        loss = criterion(outputs, labels)

        loss.backward()
        optimizer.step()

        total_loss += loss.item()
        preds = torch.argmax(outputs, dim=1)
        all_preds.extend(preds.cpu().tolist())
        all_labels.extend(labels.cpu().tolist())

    # Calculate metrics
    train_acc, train_f1 = calculate_metrics(torch.tensor(all_preds),
                                           torch.tensor(all_labels))

    # Validation
    model.eval()
    val_preds = []
    val_labels = []
    with torch.no_grad():
        for batch in test_loader:
            inputs = batch['input_ids'].to(device)
            adj = batch['adj_matrix'].float().to(device)
            srd = batch['srd_weights'].float().to(device)
            mask = batch['aspect_mask'].float().to(device)
            labels = batch['labels'].to(device)
            alpha_weights = batch['alpha_weights'].float().to(device) # Include alpha_weights in validation

            outputs = model(inputs, adj, srd, mask, alpha_weights) # Pass alpha_weights here
            preds = torch.argmax(outputs, dim=1)
            val_preds.extend(preds.cpu().tolist())
            val_labels.extend(labels.cpu().tolist())

    val_acc, val_f1 = calculate_metrics(torch.tensor(val_preds),
                                       torch.tensor(val_labels))

    if val_f1 > best_val_f1:
        best_val_f1 = val_f1
        counter = 0
        torch.save(model.state_dict(), "best_model.pth")
    else:
        counter += 1
        if counter >= patience:
            print("Early stopping triggered!")
            break
    # Save best model
    if val_f1 > best_val_f1:
        best_val_f1 = val_f1
        torch.save(model.state_dict(), "best_model.pth")

    print(f"Epoch {epoch+1}")
    print(f"Train Loss: {total_loss/len(train_loader):.4f} | Acc: {train_acc:.2%} | F1: {train_f1:.4f}")
    print(f"Val Acc: {val_acc:.2%} | F1: {val_f1:.4f}\n")
```
